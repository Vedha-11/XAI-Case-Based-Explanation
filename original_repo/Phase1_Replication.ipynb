```python
import torch
import torch.nn as nn
from torchvision import models

DEVICE = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

class NetC(nn.Module):
    def __init__(self, net):
        super(NetC, self).__init__()

        # Remove ResNet's average pooling and 1000-class classifier
        self.main = nn.Sequential(*list(net.children()))[:-2]

        # ResNet-34 final feature map: 512 x 7 x 7
        self.avgpool = nn.AvgPool2d(7)

        # CUB-200 has 200 classes
        self.linear = nn.Linear(512, 200)

    def forward(self, imgs):
        C = self.main(imgs)

        x = self.avgpool(C)
        x = x.view(x.shape[0], x.shape[1])

        logits = self.linear(x)

        return logits, x, C


# Load ImageNet-pretrained ResNet-34
resnet = models.resnet34(
    weights=models.ResNet34_Weights.IMAGENET1K_V1
)

# Convert it to the CUB-200 model
netC = NetC(resnet)

netC = netC.to(DEVICE)

print("Device:", DEVICE)
print("Classifier:", netC.linear)
```

```python
# Reset NetC to the original ImageNet-pretrained ResNet-34
# before beginning the actual replication training.

import torch
import torch.nn as nn
from torchvision import models

# Recreate pretrained ResNet-34
resnet = models.resnet34(
    weights=models.ResNet34_Weights.IMAGENET1K_V1
)

# Recreate the paper's NetC architecture
netC = NetC(resnet).to(DEVICE)

# Training configuration
NUM_EPOCHS = 500
BATCH_SIZE = 12
NUM_WORKERS = 2
INITIAL_LR = 1e-4          # REPLICATION ASSUMPTION
MOMENTUM = 0.9
WEIGHT_DECAY = 1e-5
LR_DECAY = 0.999

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.SGD(
    netC.parameters(),
    lr=INITIAL_LR,
    momentum=MOMENTUM,
    weight_decay=WEIGHT_DECAY
)

scheduler = torch.optim.lr_scheduler.MultiplicativeLR(
    optimizer,
    lr_lambda=lambda epoch: LR_DECAY
)

print("Model reset successfully.")
print("Epochs:", NUM_EPOCHS)
print("Batch size:", BATCH_SIZE)
print("Initial LR:", INITIAL_LR)
print("Momentum:", MOMENTUM)
print("Weight decay:", WEIGHT_DECAY)
print("LR decay:", LR_DECAY)
print("Device:", DEVICE)
```

```python
import os
import torch

LATEST_CHECKPOINT = os.path.join(
    CHECKPOINT_DIR,
    "cnnNORMAL_latest.pth"
)

BEST_CHECKPOINT = os.path.join(
    CHECKPOINT_DIR,
    "cnnNORMAL_best.pth"
)

start_epoch = 1
best_test_acc = 0.0

# Resume if a previous checkpoint exists
if os.path.exists(LATEST_CHECKPOINT):
    print("Loading existing checkpoint...")

    checkpoint = torch.load(
        LATEST_CHECKPOINT,
        map_location=DEVICE
    )

    netC.load_state_dict(checkpoint["model_state_dict"])
    optimizer.load_state_dict(checkpoint["optimizer_state_dict"])

    start_epoch = checkpoint["epoch"] + 1
    best_test_acc = checkpoint.get("best_test_acc", 0.0)

    print("Resuming from epoch:", start_epoch)
    print("Best test accuracy:", best_test_acc)

else:
    print("Starting training from epoch 1.")

train_losses = []
test_accuracies = []

for epoch in range(start_epoch, NUM_EPOCHS + 1):

    # -------------------------
    # Training
    # -------------------------
    netC.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        logits, _, _ = netC(images)

        loss = criterion(logits, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = logits.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    train_loss = running_loss / total
    train_acc = 100.0 * correct / total

    # -------------------------
    # Testing
    # -------------------------
    netC.eval()

    test_correct = 0
    test_total = 0

    with torch.no_grad():

        for images, labels in test_loader:

            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            logits, _, _ = netC(images)

            predictions = logits.argmax(dim=1)

            test_correct += (predictions == labels).sum().item()
            test_total += labels.size(0)

    test_acc = 100.0 * test_correct / test_total

    # -------------------------
    # Learning-rate decay
    # -------------------------
    scheduler.step()

    current_lr = optimizer.param_groups[0]["lr"]

    train_losses.append(train_loss)
    test_accuracies.append(test_acc)

    print(
        f"Epoch [{epoch:03d}/{NUM_EPOCHS}] | "
        f"Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.2f}% | "
        f"Test Acc: {test_acc:.2f}% | "
        f"LR: {current_lr:.8f}"
    )

    # -------------------------
    # Save latest checkpoint
    # -------------------------
    torch.save({
        "epoch": epoch,
        "model_state_dict": netC.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "best_test_acc": best_test_acc,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "test_acc": test_acc,
    }, LATEST_CHECKPOINT)

    # -------------------------
    # Save best model
    # -------------------------
    if test_acc > best_test_acc:

        best_test_acc = test_acc

        torch.save(
            netC.state_dict(),
            BEST_CHECKPOINT
        )

        print(
            f"  → New best model saved: "
            f"{best_test_acc:.2f}%"
        )

print("\nTraining completed.")
print("Best test accuracy: {:.2f}%".format(best_test_acc))
```

```python
import torch
import torch.nn as nn

class NetC(nn.Module):
    def __init__(self, net):
        super(NetC, self).__init__()

        self.main = nn.Sequential(
            *list(net.children())
        )[:-2]

        self.avgpool = nn.AvgPool2d(7)

        self.linear = nn.Linear(
            512,
            200
        )

    def forward(self, imgs):
        C = self.main(imgs)

        x = self.avgpool(C)

        x = x.view(
            x.shape[0],
            x.shape[1]
        )

        logits = self.linear(x)

        return logits, x, C

print("NetC class defined successfully.")
```

```python
# ============================================================
# EXPERIMENT 1 — SINGLE QUERY SANITY TEST
# ============================================================

import os
import numpy as np
import torch
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

DEVICE = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

BATCH_SIZE = 1
IMG_SIZE = 224
ALPHA = 5
FILTER_SIZE = 7
SEGMENT_DIVISIONS = 30

print("Device:", DEVICE)

# ------------------------------------------------------------
# Load reconstructed arrays
# ------------------------------------------------------------

DATA_DIR = os.path.join(CUB_DIR, "data")

X_train_c = np.load(os.path.join(DATA_DIR, "X_train_cont.npy"))
X_test_c = np.load(os.path.join(DATA_DIR, "X_test_cont.npy"))

train_preds = np.load(os.path.join(DATA_DIR, "X_train_y.npy"))
test_preds = np.load(os.path.join(DATA_DIR, "X_test_y.npy"))

# ------------------------------------------------------------
# KNN — same configuration as repository
# ------------------------------------------------------------

twin = KNeighborsClassifier(
    n_neighbors=1,
    algorithm="brute",
    metric="euclidean"
)

twin.fit(X_train_c, train_preds)

# ------------------------------------------------------------
# Load best verified model
# ------------------------------------------------------------

saved_netC = saved_netC.to(DEVICE)
saved_netC.eval()

# Classifier weights used by CAM
WEIGHTS = saved_netC.linear.weight.detach()

print("Model device:", next(saved_netC.parameters()).device)
print("WEIGHTS shape:", WEIGHTS.shape)

# ------------------------------------------------------------
# Get first test query
# ------------------------------------------------------------

QUERY_IDX = 0

QUERY_IMG, QUERY_LABEL = test_loader.dataset[QUERY_IDX]

QUERY_IMG = QUERY_IMG.unsqueeze(0).to(DEVICE)
QUERY_LABEL = torch.tensor(QUERY_LABEL).to(DEVICE)

QUERY_PRED = int(test_preds[QUERY_IDX])

QUERY_CONT = X_test_c[QUERY_IDX]

# ------------------------------------------------------------
# CNN prediction
# ------------------------------------------------------------

with torch.no_grad():
    logits, query_x, query_C = saved_netC(QUERY_IMG)

QUERY_LOGIT = logits[0][QUERY_PRED].item()
QUERY_PROB = torch.softmax(logits, dim=1)[0][QUERY_PRED]

ACTUAL_MODEL_PRED = torch.argmax(logits, dim=1).item()

# ------------------------------------------------------------
# Retrieve 10 nearest neighbors
# ------------------------------------------------------------

XP_IDXS = twin.kneighbors(
    X=[QUERY_CONT],
    n_neighbors=10,
    return_distance=False
)

NEIGHBORS = XP_IDXS[0]

# ------------------------------------------------------------
# CAM
# ------------------------------------------------------------

def get_cam_simple(C, pred):
    """
    Repository CAM implementation:
    classifier weights × convolutional feature maps,
    summed over feature channels.
    """
    if len(C.shape) == 4:
        C = C[0]

    test_weights = WEIGHTS[pred]

    C_conts = (
        test_weights.reshape(-1, 1, 1) * C
    )

    return C_conts.sum(axis=0)


cam = get_cam_simple(
    query_C,
    ACTUAL_MODEL_PRED
)

# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

print("\n==============================")
print("SINGLE QUERY SANITY TEST")
print("==============================")

print("Query index       :", QUERY_IDX)
print("True label        :", QUERY_LABEL.item())
print("Stored prediction :", QUERY_PRED)
print("Actual prediction :", ACTUAL_MODEL_PRED)
print("Query logit       :", QUERY_LOGIT)
print("Query probability :", QUERY_PROB.item())

print("\nRepresentations:")
print("query_x shape     :", query_x.shape)
print("query_C shape     :", query_C.shape)
print("CAM shape         :", cam.shape)

print("\nNearest neighbors:")
print("Indices            :", NEIGHBORS)
print("Labels             :", train_preds[NEIGHBORS])

print("\nCAM:")
print("Min                :", cam.min().item())
print("Max                :", cam.max().item())
print("Mean               :", cam.mean().item())

print("\nSingle-query setup completed.")
```

```python
# ============================================================
# Repository Experiment 1 CAM / FAM / Random functions
# ============================================================

def get_cam(C, pred, cam_type='CAM'):

    if len(C.shape) == 4:
        C = C[0]

    # -------------------------
    # CAM
    # -------------------------
    if cam_type == 'CAM':

        test_weights = WEIGHTS[pred]

        C_conts = (
            test_weights.reshape(
                test_weights.shape[0], 1, 1
            ) * C
        )

        return C_conts.sum(axis=0)

    # -------------------------
    # FAM
    # -------------------------
    if cam_type == 'FAM':

        gap = torch.nn.AvgPool2d(7)

        x = gap(C).view(-1)

        test_weights = WEIGHTS[pred]

        c = x * test_weights

        nb_feature_idx = torch.argmax(c).item()

        cam = C[nb_feature_idx]

        return cam

    # -------------------------
    # Random
    # -------------------------
    if cam_type == 'Random':

        cam = torch.tensor(
            np.random.rand(7, 7),
            dtype=torch.float32
        )

        return cam

    raise ValueError(f"Unknown cam_type: {cam_type}")


print("get_cam() defined successfully.")
```

```python
# ============================================================
# FAST FIGURE 2(A) REPLICATION
#
# Existing:
#   SLIC 30 -> already completed
#
# New:
#   SLIC 10, 20, 40, 50
#
# IMPORTANT:
#   - Uses netC_rep
#   - Does NOT retrain
#   - Does NOT execute original testing.py
#   - Does NOT calculate CCR distances
#   - Batches all superpixel candidates on GPU
#   - Saves progress continuously
# ============================================================

import os
import gc
import time
from pathlib import Path

import numpy as np
import pandas as pd
import torch

from skimage.segmentation import slic


# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

os.chdir(str(CUB_DIR))

FAST_DIR = CUB_DIR / "data" / "NewTestingResults" / "FastFigure2A"
FAST_DIR.mkdir(parents=True, exist_ok=True)

print("Output directory:")
print(FAST_DIR)


# ------------------------------------------------------------
# Settings
# ------------------------------------------------------------

SEGMENT_SETTINGS = [10, 20, 40, 50]

START_QUERY = 0
END_QUERY = 500

SAVE_EVERY = 10

IMAGE_PIXELS = 224 * 224


# ------------------------------------------------------------
# Make sure model is in evaluation mode
# ------------------------------------------------------------

netC_rep.eval()


# ------------------------------------------------------------
# Helper: run model
# ------------------------------------------------------------

@torch.inference_mode()
def model_logits(batch):
    return netC_rep(batch)[0]


# ------------------------------------------------------------
# Fast experiment for ONE segmentation setting
# ------------------------------------------------------------

def run_fast_setting(seg_div):

    output_file = FAST_DIR / f"Figure2A_Fast_Segments_{seg_div}.csv"

    # --------------------------------------------------------
    # Resume if partial file already exists
    # --------------------------------------------------------

    if output_file.exists():

        existing = pd.read_csv(output_file)

        if len(existing) > 0:
            completed_queries = set(
                existing["InstanceIdx"].astype(int).unique()
            )

            print(
                f"\nSLIC {seg_div}: "
                f"found existing file with "
                f"{len(completed_queries)} queries."
            )

        else:
            existing = pd.DataFrame()
            completed_queries = set()

    else:
        existing = pd.DataFrame()
        completed_queries = set()


    rows = []

    if len(existing) > 0:
        rows.extend(existing.to_dict("records"))


    # --------------------------------------------------------
    # Process queries
    # --------------------------------------------------------

    for query_idx in range(START_QUERY, END_QUERY + 1):

        if query_idx in completed_queries:
            continue

        start_time = time.time()

        # ----------------------------------------------------
        # Load image
        # ----------------------------------------------------

        img, label = test_dataset[query_idx]

        QUERY_IMG = img.unsqueeze(0).to(DEVICE)

        # ----------------------------------------------------
        # Original prediction
        # ----------------------------------------------------

        with torch.inference_mode():

            logits, query_cont, _ = netC_rep(QUERY_IMG)

            QUERY_PRED = torch.argmax(
                logits,
                dim=1
            ).item()

            QUERY_LOGIT = logits[
                0,
                QUERY_PRED
            ].item()

            QUERY_PROB = torch.softmax(
                logits,
                dim=1
            )[0, QUERY_PRED]


        # ----------------------------------------------------
        # Convert normalized tensor back to image
        # ----------------------------------------------------

        image = (
            repo_testing_local.inverse_normalize(
                QUERY_IMG,
                mean=repo_testing_local.MEAN_NORM,
                std=repo_testing_local.STD_NORM
            ) * 255
        )[0].permute(
            1, 2, 0
        ).cpu().numpy().round().astype("uint8")


        # ----------------------------------------------------
        # SLIC
        # ----------------------------------------------------

        segments = slic(
            image,
            n_segments=seg_div,
            sigma=5,
            start_label=1
        )

        unique_segments = np.unique(segments)


        # ====================================================
        # STEP 1
        # Find the most salient superpixel
        #
        # IMPORTANT:
        # Evaluate ALL superpixels together in one GPU batch.
        # ====================================================

        candidate_images = []

        candidate_ids = []

        for sp_id in unique_segments:

            # Same operation used by repository
            isolated = repo_testing_local.keep_ccr_superpixel(
                image,
                segments,
                sp_id,
                []
            )

            pil_image = repo_testing_local.Image.fromarray(
                isolated
            )

            tensor_image = (
                repo_testing_local.transformNormalize(
                    pil_image
                )
            )

            candidate_images.append(
                tensor_image
            )

            candidate_ids.append(
                int(sp_id)
            )


        # Stack all candidate superpixels
        candidate_batch = torch.stack(
            candidate_images,
            dim=0
        ).to(DEVICE)


        # ----------------------------------------------------
        # One batched forward pass
        # ----------------------------------------------------

        with torch.inference_mode():

            candidate_logits = model_logits(
                candidate_batch
            )

            candidate_class_logits = (
                candidate_logits[:, QUERY_PRED]
            )

        # Most salient superpixel = highest original-class logit
        best_position = torch.argmax(
            candidate_class_logits
        ).item()

        best_sp_id = candidate_ids[
            best_position
        ]


        # ----------------------------------------------------
        # Pixel count of selected superpixel
        # ----------------------------------------------------

        pixel_count = int(
            (segments == best_sp_id).sum()
        )

        area_percent = (
            pixel_count /
            IMAGE_PIXELS *
            100
        )


        # ====================================================
        # STEP 2
        # Superpixel occlusion
        # ====================================================

        masked_sp = (
            repo_testing_local.blackout_ccr_superpixel(
                QUERY_IMG.clone(),
                segments,
                best_sp_id,
                occlude_pos=True
            )
        )

        with torch.inference_mode():

            sp_logits = model_logits(
                masked_sp
            )

        sp_logit = sp_logits[
            0,
            QUERY_PRED
        ].item()

        sp_drop = QUERY_LOGIT - sp_logit


        # ====================================================
        # STEP 3
        # CAM / FAM / Random
        #
        # Same pixel count as superpixel.
        # ====================================================

        method_results = {
            "Superpixel": sp_drop
        }


        for method in [
            "CAM",
            "FAM",
            "Random"
        ]:

            # Repository implementation
            repo_testing_local.QUERY_IMG = QUERY_IMG
            repo_testing_local.QUERY_PRED = QUERY_PRED
            repo_testing_local.QUERY_IDX = query_idx
            repo_testing_local.QUERY_CONT = query_cont
            repo_testing_local.QUERY_LOGIT = QUERY_LOGIT
            repo_testing_local.QUERY_PROB = QUERY_PROB

            cam = (
                repo_testing_local
                .get_upsampled_cam_query(
                    netC_rep,
                    cam_type=method
                )
            )

            masked = (
                repo_testing_local
                .get_cam_image_masked(
                    QUERY_IMG.clone(),
                    cam,
                    pixel_count,
                    occlude_pos=True
                )
            )

            with torch.inference_mode():

                method_logits = model_logits(
                    masked
                )

            method_logit = method_logits[
                0,
                QUERY_PRED
            ].item()

            method_results[method] = (
                QUERY_LOGIT -
                method_logit
            )


        # ====================================================
        # Save four methods
        # ====================================================

        for method in [
            "Superpixel",
            "CAM",
            "FAM",
            "Random"
        ]:

            rows.append({
                "Method": method,
                "PixelsOccluded": pixel_count,
                "LogitDrop": method_results[method],
                "QueryLogit": QUERY_LOGIT,
                "InstanceIdx": query_idx,
                "NumSpOccluded": 1,
                "TotalSPs": len(unique_segments),
                "SLICSegments": seg_div,
                "AreaPercent": area_percent
            })


        # ----------------------------------------------------
        # Periodic checkpoint
        # ----------------------------------------------------

        if (
            (query_idx + 1) % SAVE_EVERY == 0
            or query_idx == END_QUERY
        ):

            pd.DataFrame(rows).to_csv(
                output_file,
                index=False
            )

            torch.cuda.empty_cache()
            gc.collect()


        # ----------------------------------------------------
        # Progress
        # ----------------------------------------------------

        elapsed = time.time() - start_time

        print(
            f"SLIC={seg_div:2d} | "
            f"Query={query_idx:3d}/500 | "
            f"SPs={len(unique_segments):2d} | "
            f"Pixels={pixel_count:5d} | "
            f"Time={elapsed:.2f}s"
        )


    # --------------------------------------------------------
    # Final save
    # --------------------------------------------------------

    final_df = pd.DataFrame(rows)

    final_df.to_csv(
        output_file,
        index=False
    )

    print(
        f"\nCompleted SLIC {seg_div}"
    )

    print(
        f"Saved: {output_file}"
    )

    return final_df


# ============================================================
# RUN ALL MISSING SETTINGS
# ============================================================

all_fast_results = {}

for seg_div in SEGMENT_SETTINGS:

    print("\n")
    print("=" * 80)
    print(f"STARTING FAST SLIC = {seg_div}")
    print("=" * 80)

    result = run_fast_setting(seg_div)

    all_fast_results[seg_div] = result


print("\n")
print("=" * 80)
print("ALL FOUR SETTINGS COMPLETED")
print("=" * 80)

for seg_div, result in all_fast_results.items():

    print(
        f"SLIC {seg_div}: "
        f"{len(result)} rows"
    )
```

```python
# ============================================================
# FAST FIGURE 2(C) — CUB-200 INCLUSION
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import torch
from PIL import Image

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------
CUB_DIR = Path(
    r"C:\Users\Lenovo\OneDrive\Desktop\XAI-Case-Based-Explanation\original_repo\Expt 1\CUB"
)

INCLUSION_DIR = (
    CUB_DIR
    / "data"
    / "NewTestingResults"
    / "FastFigure2C"
)

INCLUSION_DIR.mkdir(parents=True, exist_ok=True)

SEGMENT_SETTINGS = [10, 20, 40, 50]

NUM_QUERIES = 500

# ------------------------------------------------------------
# Helper: model logits
# ------------------------------------------------------------
@torch.no_grad()
def model_logits(images):
    if images.dim() == 3:
        images = images.unsqueeze(0)

    logits, _, _ = netC_rep(
        images.to(DEVICE)
    )

    return logits


# ------------------------------------------------------------
# Helper: create isolated superpixel
# ------------------------------------------------------------
def get_isolated_superpixel(image_np, segments, segment_id):
    isolated = repo_testing_local.keep_ccr_superpixel(
        image_np,
        segments,
        segment_id,
        []
    )
    return isolated


# ------------------------------------------------------------
# Helper: create inclusion image
# Keep selected pixels, black out everything else.
# ------------------------------------------------------------
def inclusion_mask_image(image_tensor, saliency_map, pixel_count):
    
    # Copy image
    img = image_tensor.clone()

    # Find the highest-saliency pixels
    idx, idy = repo_testing_local.largest_indices(
        saliency_map,
        pixel_count
    )

    # Start with everything removed
    mask = torch.zeros(
        saliency_map.shape,
        device=DEVICE
    )

    # Keep only the selected pixels
    for x, y in zip(idx, idy):
        mask[x][y] = 1

    img = img * mask

    # Match the normalization used by the repository
    img[img == 0] = -2.1179

    return img


# ------------------------------------------------------------
# Run one segmentation setting
# ------------------------------------------------------------
def run_inclusion_setting(seg_div, max_queries=500):

    output_file = (
        INCLUSION_DIR /
        f"Figure2C_Fast_Segments_{seg_div}.csv"
    )

    # Resume if already complete
    if output_file.exists():

        existing = pd.read_csv(output_file)

        expected = max_queries * 4

        if len(existing) >= expected:

            print(
                f"SLIC {seg_div}: already complete "
                f"({len(existing)} rows)"
            )

            return existing

    rows = []

    print("\n" + "=" * 80)
    print(f"STARTING CUB INCLUSION — SLIC {seg_div}")
    print("=" * 80)

    for query_idx in range(max_queries):

        # ----------------------------------------------------
        # Load query image
        # ----------------------------------------------------
        query_img, query_label = test_dataset[query_idx]

        query_img = query_img.to(DEVICE)

        # Original prediction
        with torch.no_grad():

            query_logits = model_logits(
                query_img
            )

        query_pred = int(
            torch.argmax(query_logits[0]).item()
        )

        query_logit = float(
            query_logits[0][query_pred].item()
        )

        # ----------------------------------------------------
        # Convert image back to pixel space
        # ----------------------------------------------------
        image = (
            repo_testing_local.inverse_normalize(
                query_img.unsqueeze(0),
                repo_testing_local.MEAN_NORM,
                repo_testing_local.STD_NORM
            )[0]
            .permute(1, 2, 0)
            .detach()
            .cpu()
            .numpy()
            * 255
        )

        image = np.clip(
            image,
            0,
            255
        ).astype("uint8")

        # ----------------------------------------------------
        # SLIC segmentation
        # ----------------------------------------------------
        segments = repo_testing_local.slic(
            image,
            n_segments=seg_div,
            sigma=5,
            start_label=1
        )

        segment_ids = np.unique(segments)

        # ----------------------------------------------------
        # Create all isolated superpixels
        # ----------------------------------------------------
        candidate_tensors = []
        candidate_ids = []

        for segment_id in segment_ids:

            isolated = get_isolated_superpixel(
                image,
                segments,
                segment_id
            )

            pil_image = Image.fromarray(
                isolated
            )

            tensor_img = (
                repo_testing_local.transformNormalize(
                    pil_image
                )
            )

            candidate_tensors.append(
                tensor_img
            )

            candidate_ids.append(
                int(segment_id)
            )

        # ----------------------------------------------------
        # Batch forward pass
        # Find most salient superpixel by original
        # predicted-class logit.
        # ----------------------------------------------------
        candidate_batch = torch.stack(
            candidate_tensors
        ).to(DEVICE)

        with torch.no_grad():

            candidate_logits = model_logits(
                candidate_batch
            )

        candidate_scores = (
            candidate_logits[:, query_pred]
        )

        best_position = int(
            torch.argmax(candidate_scores).item()
        )

        best_segment_id = candidate_ids[
            best_position
        ]

        # ----------------------------------------------------
        # Selected superpixel mask
        # ----------------------------------------------------
        selected_mask = (
            segments == best_segment_id
        )

        pixel_count = int(
            selected_mask.sum()
        )

        area_percent = (
            pixel_count /
            (224 * 224)
            * 100
        )

        # ----------------------------------------------------
        # SUPERPIXEL INCLUSION
        # ----------------------------------------------------
        isolated_best = get_isolated_superpixel(
            image,
            segments,
            best_segment_id
        )

        pil_best = Image.fromarray(
            isolated_best
        )

        superpixel_tensor = (
            repo_testing_local.transformNormalize(
                pil_best
            )
        )

        with torch.no_grad():

            sp_logits = model_logits(
                superpixel_tensor
            )

        sp_logit = float(
            sp_logits[0, query_pred].item()
        )

        # ----------------------------------------------------
        # CAM / FAM / RANDOM
        # Equal pixel area to selected superpixel
        # ----------------------------------------------------
        for method in [
            "CAM",
            "FAM",
            "Random"
        ]:

            cam = (
                repo_testing_local.get_upsampled_cam_query(
                    query_img,
                    query_pred,
                    method
                )
            )

            # IMPORTANT:
            # For inclusion we keep the most salient pixels
            # rather than occluding them.
            included = inclusion_mask_image(
                query_img,
                cam,
                pixel_count
            )

            with torch.no_grad():

                method_logits = model_logits(
                    included
                )

            method_logit = float(
                method_logits[0, query_pred].item()
            )

            rows.append(
                {
                    "Method": method,
                    "PixelsIncluded": pixel_count,
                    "Logit": method_logit,
                    "QueryLogit": query_logit,
                    "InstanceIdx": query_idx,
                    "NumSpIncluded": 1,
                    "TotalSPs": len(segment_ids),
                    "SLICSegments": seg_div,
                    "AreaPercent": area_percent
                }
            )

        # ----------------------------------------------------
        # Superpixel row
        # ----------------------------------------------------
        rows.append(
            {
                "Method": "Superpixel",
                "PixelsIncluded": pixel_count,
                "Logit": sp_logit,
                "QueryLogit": query_logit,
                "InstanceIdx": query_idx,
                "NumSpIncluded": 1,
                "TotalSPs": len(segment_ids),
                "SLICSegments": seg_div,
                "AreaPercent": area_percent
            }
        )

        # ----------------------------------------------------
        # Progress
        # ----------------------------------------------------
        if query_idx % 10 == 0:

            print(
                f"SLIC {seg_div}: "
                f"Query {query_idx + 1}/{max_queries}"
            )

            # Checkpoint
            pd.DataFrame(rows).to_csv(
                output_file,
                index=False
            )

    # --------------------------------------------------------
    # Final save
    # --------------------------------------------------------
    result = pd.DataFrame(rows)

    result.to_csv(
        output_file,
        index=False
    )

    print(
        f"\nCompleted SLIC {seg_div}"
    )

    print(
        f"Saved: {output_file}"
    )

    print(
        f"Rows: {len(result)}"
    )

    return result


# ============================================================
# RUN ALL FOUR MISSING SETTINGS
# ============================================================

all_results = {}

for seg in SEGMENT_SETTINGS:

    result = run_inclusion_setting(
        seg,
        max_queries=NUM_QUERIES
    )

    all_results[seg] = result


print("\n" + "=" * 80)
print("ALL FOUR CUB INCLUSION SETTINGS COMPLETED")
print("=" * 80)

for seg, result in all_results.items():

    print(
        f"SLIC {seg}: "
        f"{len(result)} rows"
    )
```

```python
# ============================================================
# FAST FIGURE 2(C) — CUB-200 INCLUSION
# CORRECTED VERSION
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import torch
from PIL import Image

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------
CUB_DIR = Path(
    r"C:\Users\Lenovo\OneDrive\Desktop\XAI-Case-Based-Explanation\original_repo\Expt 1\CUB"
)

INCLUSION_DIR = (
    CUB_DIR
    / "data"
    / "NewTestingResults"
    / "FastFigure2C"
)

INCLUSION_DIR.mkdir(parents=True, exist_ok=True)

SEGMENT_SETTINGS = [10, 20, 30, 40, 50]
NUM_QUERIES = 500


# ============================================================
# MODEL HELPER
# ============================================================

@torch.no_grad()
def model_logits(images):

    if images.dim() == 3:
        images = images.unsqueeze(0)

    logits, _, _ = netC_rep(
        images.to(DEVICE)
    )

    return logits


# ============================================================
# INCLUSION MASK
# Keep highest-saliency pixels
# ============================================================

def inclusion_mask_image(
    image_tensor,
    saliency_map,
    pixel_count
):

    img = image_tensor.clone()

    idx, idy = repo_testing_local.largest_indices(
        saliency_map,
        pixel_count
    )

    mask = torch.zeros(
        saliency_map.shape,
        device=DEVICE
    )

    for x, y in zip(idx, idy):
        mask[x][y] = 1

    img = img * mask

    # Same normalized background value used by repository
    img[img == 0] = -2.1179

    return img


# ============================================================
# RUN ONE INCLUSION SETTING
# ============================================================

def run_inclusion_setting(
    seg_div,
    max_queries=500
):

    output_file = (
        INCLUSION_DIR /
        f"Figure2C_Fast_Segments_{seg_div}.csv"
    )

    # --------------------------------------------------------
    # Resume if complete
    # --------------------------------------------------------
    if output_file.exists():

        existing = pd.read_csv(
            output_file
        )

        expected = max_queries * 4

        if len(existing) >= expected:

            print(
                f"SLIC {seg_div}: already complete "
                f"({len(existing)} rows)"
            )

            return existing

    rows = []

    print("\n" + "=" * 80)
    print(
        f"STARTING CUB INCLUSION — SLIC {seg_div}"
    )
    print("=" * 80)

    # --------------------------------------------------------
    # Query loop
    # --------------------------------------------------------
    for query_idx in range(max_queries):

        # ----------------------------------------------------
        # Load query
        # ----------------------------------------------------
        query_img, query_label = test_dataset[
            query_idx
        ]

        query_img = query_img.to(DEVICE)

        # ----------------------------------------------------
        # Original prediction
        # ----------------------------------------------------
        with torch.no_grad():

            query_logits = model_logits(
                query_img
            )

        query_pred = int(
            torch.argmax(
                query_logits[0]
            ).item()
        )

        query_logit = float(
            query_logits[
                0,
                query_pred
            ].item()
        )

        # ----------------------------------------------------
        # Convert normalized image to pixel image
        # ----------------------------------------------------
        image = (
            repo_testing_local.inverse_normalize(
                query_img.unsqueeze(0),
                repo_testing_local.MEAN_NORM,
                repo_testing_local.STD_NORM
            )[0]
            .permute(1, 2, 0)
            .detach()
            .cpu()
            .numpy()
            * 255
        )

        image = np.clip(
            image,
            0,
            255
        ).astype("uint8")

        # ----------------------------------------------------
        # SLIC
        # ----------------------------------------------------
        segments = repo_testing_local.slic(
            image,
            n_segments=seg_div,
            sigma=5,
            start_label=1
        )

        segment_ids = np.unique(
            segments
        )

        # ----------------------------------------------------
        # Isolate every superpixel
        # ----------------------------------------------------
        candidate_tensors = []
        candidate_ids = []

        for segment_id in segment_ids:

            isolated = (
                repo_testing_local.keep_ccr_superpixel(
                    image,
                    segments,
                    segment_id,
                    []
                )
            )

            pil_image = Image.fromarray(
                isolated
            )

            tensor_img = (
                repo_testing_local.transformNormalize(
                    pil_image
                )
            )

            candidate_tensors.append(
                tensor_img
            )

            candidate_ids.append(
                int(segment_id)
            )

        # ----------------------------------------------------
        # Batch inference
        # Find most salient superpixel
        # ----------------------------------------------------
        candidate_batch = torch.stack(
            candidate_tensors
        ).to(DEVICE)

        with torch.no_grad():

            candidate_logits = model_logits(
                candidate_batch
            )

        candidate_scores = (
            candidate_logits[
                :,
                query_pred
            ]
        )

        best_position = int(
            torch.argmax(
                candidate_scores
            ).item()
        )

        best_segment_id = (
            candidate_ids[
                best_position
            ]
        )

        # ----------------------------------------------------
        # Selected superpixel
        # ----------------------------------------------------
        selected_mask = (
            segments == best_segment_id
        )

        pixel_count = int(
            selected_mask.sum()
        )

        area_percent = (
            pixel_count /
            (224 * 224)
            * 100
        )

        # ====================================================
        # SUPERPIXEL INCLUSION
        # ====================================================

        isolated_best = (
            repo_testing_local.keep_ccr_superpixel(
                image,
                segments,
                best_segment_id,
                []
            )
        )

        pil_best = Image.fromarray(
            isolated_best
        )

        superpixel_tensor = (
            repo_testing_local.transformNormalize(
                pil_best
            )
        )

        with torch.no_grad():

            sp_logits = model_logits(
                superpixel_tensor
            )

        sp_logit = float(
            sp_logits[
                0,
                query_pred
            ].item()
        )

        rows.append(
            {
                "Method": "Superpixel",
                "PixelsIncluded": pixel_count,
                "Logit": sp_logit,
                "QueryLogit": query_logit,
                "InstanceIdx": query_idx,
                "NumSpIncluded": 1,
                "TotalSPs": len(segment_ids),
                "SLICSegments": seg_div,
                "AreaPercent": area_percent
            }
        )

        # ====================================================
        # CAM / FAM / RANDOM
        # ====================================================

        # Repository function uses global query variables.
        repo_testing_local.QUERY_IMG = (
            query_img.unsqueeze(0)
        )

        repo_testing_local.QUERY_PRED = (
            query_pred
        )

        for method in [
            "CAM",
            "FAM",
            "Random"
        ]:

            # Correct repository function signature:
            # get_upsampled_cam_query(
            #     netC,
            #     cam_type,
            #     upsample
            # )
            cam = (
                repo_testing_local
                .get_upsampled_cam_query(
                    netC_rep,
                    method,
                    True
                )
            )

            # Keep exactly the same number of pixels
            # as the selected superpixel.
            included = inclusion_mask_image(
                query_img,
                cam,
                pixel_count
            )

            with torch.no_grad():

                method_logits = model_logits(
                    included
                )

            method_logit = float(
                method_logits[
                    0,
                    query_pred
                ].item()
            )

            rows.append(
                {
                    "Method": method,
                    "PixelsIncluded": pixel_count,
                    "Logit": method_logit,
                    "QueryLogit": query_logit,
                    "InstanceIdx": query_idx,
                    "NumSpIncluded": 1,
                    "TotalSPs": len(segment_ids),
                    "SLICSegments": seg_div,
                    "AreaPercent": area_percent
                }
            )

        # ----------------------------------------------------
        # Progress + checkpoint
        # ----------------------------------------------------
        if query_idx % 10 == 0:

            print(
                f"SLIC {seg_div}: "
                f"Query {query_idx + 1}/{max_queries}"
            )

            pd.DataFrame(
                rows
            ).to_csv(
                output_file,
                index=False
            )

    # --------------------------------------------------------
    # Final save
    # --------------------------------------------------------
    result = pd.DataFrame(
        rows
    )

    result.to_csv(
        output_file,
        index=False
    )

    print(
        f"\nCompleted SLIC {seg_div}"
    )

    print(
        f"Saved: {output_file}"
    )

    print(
        f"Rows: {len(result)}"
    )

    return result


# ============================================================
# RUN ALL FOUR SETTINGS
# ============================================================

all_results = {}

for seg in SEGMENT_SETTINGS:

    result = run_inclusion_setting(
        seg,
        max_queries=NUM_QUERIES
    )

    all_results[seg] = result


print("\n" + "=" * 80)
print("ALL FOUR CUB INCLUSION SETTINGS COMPLETED")
print("=" * 80)

for seg, result in all_results.items():

    print(
        f"SLIC {seg}: "
        f"{len(result)} rows"
    )
```
